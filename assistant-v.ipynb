```python
import speech_recognition as sr
import pyttsx3

# Initialize the recognizer and text-to-speech engine
recognizer = sr.Recognizer()
tts_engine = pyttsx3.init()

def speak(text):
    """Convert text to speech."""
    tts_engine.say(text)
    tts_engine.runAndWait()

def listen():
    """Capture audio from the microphone and convert it to text."""
    with sr.Microphone() as source:
        print("Listening...")
        audio = recognizer.listen(source)
        try:
            text = recognizer.recognize_google(audio)
            return text
        except sr.UnknownValueError:
            return "Sorry, I did not understand that."
        except sr.RequestError:
            return "Sorry, my speech service is down."

def handle_command(command):
    """Handle the recognized command."""
    command = command.lower()
    if "hello" in command:
        response = "Hello! How can I help you today?"
    elif "how are you" in command:
        response = "I'm an AI, so I don't have feelings, but thanks for asking!"
    elif "your name" in command:
        response = "I am your voice assistant."
    elif "time" in command:
        from datetime import datetime
        now = datetime.now().strftime("%H:%M")
        response = f"The current time is {now}."
    elif "exit" in command or "quit" in command:
        response = "Goodbye!"
        speak(response)
        exit()
    elif "weather" in command:
        # Integrate weather API here
        response = "Currently, I cannot fetch the weather for you."

    elif "news" in command:
        # Integrate news API here
        response = "Currently, I cannot fetch the news for you."
    elif "reminder" in command:
        # Add reminder functionality here
        response = "I can set a reminder for you."
    elif "plus" in command or "minus" in command or "multiply" in command or "divide" in command:
        try:
            response = str(eval(command))
        except:
            response = "Sorry, I couldn't compute that."
    else:
        response = "Sorry, I can't help with that."

    return response

def main():
    """Main function to run the voice assistant."""
    while True:
        command = listen()
        print(f"You said: {command}")
        response = handle_command(command)
        print(f"Assistant: {response}")
        speak(response)

if __name__ == "__main__":
    main()
```
